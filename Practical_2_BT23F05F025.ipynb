# 02. Probabilistic Modeling and Synthetic Data

## Aim

To implement probabilistic modeling techniques and generate synthetic data samples using a Gaussian probability distribution.

## Theory

Probabilistic modeling is a technique used to represent and understand data using probability distributions. It helps in modeling uncertainty and generating new data samples that follow a similar statistical pattern.

### Gaussian Distribution

The Gaussian distribution, also known as the Normal distribution, is one of the most commonly used probability distributions.

It is characterized by two parameters:

- Mean (μ): Represents the center of the distribution.
- Standard Deviation (σ): Represents the spread of the distribution.

The Gaussian distribution has a bell-shaped curve and is symmetric around its mean.

### Sampling

Sampling is the process of generating observations from a probability distribution.

In this practical, synthetic data is generated using a Gaussian distribution with:

- Mean = 50
- Standard Deviation = 10
- Number of samples = 1000

### Probability Density Function

The Probability Density Function (PDF) describes the relative likelihood of a continuous random variable taking a particular value.

For a Gaussian distribution, the PDF produces a bell-shaped curve.

### Synthetic Data

Synthetic data is artificially generated data created using statistical or machine learning models.

In this practical, synthetic data is generated from a Gaussian distribution and compared with its theoretical PDF.

## Expected Learning

After performing this practical, we understand how probability distributions can be used to model data and generate synthetic samples.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Gaussian distribution parameters
mean = 50
std = 10

# 2. Sampling: generate synthetic data
data = np.random.normal(mean, std, 1000)

# 3. Probability Density Function
x = np.linspace(10, 90, 1000)
pdf = norm.pdf(x, mean, std)

# 4. Data visualization
plt.hist(data, bins=30, density=True, alpha=0.6, label="Synthetic Data")
plt.plot(x, pdf, 'r', label="Gaussian PDF")

plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Gaussian Distribution and Synthetic Data")
plt.legend()
plt.show()
```

## Output

The generated histogram represents 1000 synthetic data samples generated from a Gaussian distribution with mean 50 and standard deviation 10.

The red curve represents the theoretical Gaussian Probability Density Function (PDF).

### Observation

- The synthetic data is centered approximately around 50.
- Most of the generated values lie close to the mean.
- The spread of the data is controlled by the standard deviation of 10.
- The histogram approximately follows the Gaussian PDF curve.
- The result demonstrates that probabilistic models can be used to generate synthetic data.

## Conclusion

Thus, probabilistic modeling was implemented using a Gaussian distribution. Synthetic data samples were generated using specified mean and standard deviation, and the generated data was visualized along with the theoretical Gaussian PDF.
