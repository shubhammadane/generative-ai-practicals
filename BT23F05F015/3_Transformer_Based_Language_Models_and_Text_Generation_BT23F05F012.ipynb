# Practical 3: Transformer-Based Language Models and Text Generation

## Aim

To study the architecture of transformer-based language models and generate text using pretrained models.

## Introduction

Transformer is a deep learning architecture widely used in Natural Language Processing (NLP).
It uses the attention mechanism to understand relationships between words in a sequence.

Transformer-based language models such as GPT and BERT have achieved excellent performance in various NLP tasks.

In this practical, we study the basic architecture and components of Transformer-based language models and use a pretrained GPT-2 model to generate text from a given prompt.

## Transformer Architecture

The Transformer architecture was introduced to process sequential data using the attention mechanism instead of relying only on recurrent neural networks.

### Main Components of Transformer

1. Self-Attention:
   Helps the model understand the relationship between different words in a sentence.

2. Multi-Head Attention:
   Allows the model to focus on different relationships between words simultaneously.

3. Positional Encoding:
   Provides information about the position of words in the input sequence.

4. Feed Forward Neural Network:
   Processes the representations generated by the attention layers.

5. Layer Normalization:
   Helps stabilize and improve the training of the model.

6. Residual Connections:
   Help information flow through deeper layers and make training easier.

### Transformer-based Language Models

- BERT: Mainly uses the Transformer encoder and is designed for understanding text.
- GPT: Uses the Transformer decoder architecture and is designed for autoregressive text generation.

```python
from transformers import pipeline

print("Transformers library imported successfully!")
```

```python
from transformers import pipeline

# Load a pretrained GPT-2 text generation model
generator = pipeline("text-generation", model="gpt2")

print("Pretrained GPT-2 model loaded successfully!")
```

```python
# Input prompt
prompt = "Artificial Intelligence is"

# Generate text using the pretrained GPT-2 model
result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    clean_up_tokenization_spaces=False
)

print("Input Prompt:")
print(prompt)

print("\nGenerated Text:")
print(result[0]["generated_text"])
```

```python
prompts = [
    "The Technology is  ",
    "Artifical Intelligence is ",
    "AI is developing",
]

for prompt in prompts:
    result = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        clean_up_tokenization_spaces=False
    )

    print("Prompt:", prompt)
    print("Generated Text:", result[0]["generated_text"])
    print("-" * 80)
```

## GPT-2 Architecture

GPT-2 (Generative Pre-trained Transformer 2) is a Transformer-based autoregressive language model developed by OpenAI.

GPT-2 uses the Transformer decoder architecture to generate text.

### Working of GPT-2

1. The input text is first converted into tokens.
2. Tokens are converted into numerical representations.
3. Positional information is added to preserve the order of tokens.
4. Self-attention helps the model identify relationships between tokens.
5. Multiple Transformer decoder blocks process the input.
6. The model predicts the probability of the next token.
7. The predicted token is added to the sequence.
8. This process continues until the required number of tokens is generated.

### Text Generation Flow

Input Prompt
      ↓
Tokenization
      ↓
Token Embeddings + Positional Information
      ↓
Transformer Decoder Blocks
      ↓
Self-Attention
      ↓
Feed Forward Network
      ↓
Next Token Prediction
      ↓
Generated Text
