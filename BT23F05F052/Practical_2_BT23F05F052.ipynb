```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# ============================================================
# PRACTICAL 2: PROBABILISTIC MODELING & SYNTHETIC DATA
# ============================================================

# 1. Define Gaussian distribution parameters
mean = 50
std = 10

print("Mean:", mean)
print("Standard Deviation:", std)


# 2. Generate synthetic data
data = np.random.normal(mean, std, 1000)

print("\nSynthetic data generated:", len(data))
print("Sample mean:", np.mean(data))
print("Sample standard deviation:", np.std(data))


# 3. Calculate Probability Density Function (PDF)
x = np.linspace(10, 90, 1000)
pdf = norm.pdf(x, mean, std)


# 4. Visualize synthetic data and Gaussian PDF
plt.figure(figsize=(10, 5))

plt.hist(
    data,
    bins=30,
    density=True,
    alpha=0.6,
    label="Synthetic Data"
)

plt.plot(
    x,
    pdf,
    label="Gaussian PDF"
)

plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Gaussian Distribution and Synthetic Data")
plt.legend()
plt.grid(alpha=0.2)

plt.show()


# 5. Observation
print("\nObservation:")
print("The synthetic data follows approximately a Gaussian distribution.")
print("Most values are concentrated around the mean of 50.")
print("The generated data can be used as synthetic data for modeling.")


# 6. Conclusion
print("\nConclusion:")
print("Probabilistic modeling can be used to generate synthetic data")
print("based on a specified probability distribution.")
```
