```python

# PRACTICAL 6: TEXT-TO-IMAGE GENERATION USING STABLE DIFFUSION

import torch
import matplotlib.pyplot as plt
from diffusers import DiffusionPipeline

print("=" * 60)
print("PRACTICAL 6: TEXT-TO-IMAGE GENERATION")
print("=" * 60)

# Use CPU
device = "cpu"
print("Using device:", device)

# Load lightweight pretrained model
model_id = "segmind/tiny-sd"

print("\nLoading lightweight Stable Diffusion model...")

pipe = DiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float32
)

pipe = pipe.to(device)

print("Model loaded successfully!")

# Text prompt
prompt = "A beautiful futuristic city with green gardens, modern buildings, mountains, realistic digital art"

print("\nInput Prompt:")
print(prompt)

# Generate image
print("\nGenerating image...")

with torch.inference_mode():
    image = pipe(
        prompt,
        num_inference_steps=5,
        guidance_scale=5.0,
        height=256,
        width=256
    ).images[0]

# Display generated image
plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.axis("off")
plt.title("Image Generated Using Stable Diffusion")
plt.show()

# Save image
image.save("generated_image.png")

print("\nImage saved as generated_image.png")

print("\nOBSERVATION:")
print("The pretrained diffusion model generated an image based on the text prompt.")

print("\nCONCLUSION:")
print("Stable Diffusion converts text prompts into images using a pretrained diffusion model.")

print("\nPRACTICAL 6 COMPLETED")
print("=" * 60)
```
