```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

model_name = "sshleifer/tiny-gpt2"

print("=" * 60)
print("PRACTICAL 3: TRANSFORMER-BASED LANGUAGE MODELS")
print("=" * 60)

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print("\nModel loaded successfully!")

print("\nMODEL INFORMATION")
print("-" * 60)
print("Model:", model_name)
print("Vocabulary Size:", tokenizer.vocab_size)
print("Transformer Layers:", model.config.n_layer)
print("Attention Heads:", model.config.n_head)

text = "Artificial Intelligence is transforming the world."

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("\nTOKENIZATION")
print("-" * 60)
print("Original Text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
print("Decoded Text:", tokenizer.decode(token_ids))

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

prompt = "Artificial Intelligence is"

result = generator(
    prompt,
    max_new_tokens=30,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7,
    pad_token_id=tokenizer.eos_token_id
)

print("\nTEXT GENERATION")
print("-" * 60)
print("Prompt:", prompt)
print("Generated Text:", result[0]["generated_text"])

print("\nTRANSFORMER ARCHITECTURE")
print("-" * 60)
print("Input text is converted into tokens.")
print("Tokens are converted into numerical representations.")
print("Self-attention helps the model understand relationships between tokens.")
print("Transformer layers process the token representations.")
print("The model predicts the next token to generate text.")

print("\nOBSERVATION")
print("-" * 60)
print("The pretrained Transformer model successfully processed")
print("tokens and generated text from the given prompt.")

print("\nCONCLUSION")
print("-" * 60)
print("Transformer-based language models use attention mechanisms")
print("to process tokens and generate text based on input prompts.")

print("\n" + "=" * 60)
print("PRACTICAL 3 COMPLETED")
print("=" * 60)
```
