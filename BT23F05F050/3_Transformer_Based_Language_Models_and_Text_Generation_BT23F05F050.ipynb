```python
!pip install -q transformers torch

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

print("=" * 70)
print("PRACTICAL 3: TRANSFORMER-BASED LANGUAGE MODELS")
print("=" * 70)

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

print("\nGPT-2 model loaded successfully!")

print("\nMODEL INFORMATION")
print("-" * 70)
print("Model name:", model_name)
print("Vocabulary size:", tokenizer.vocab_size)
print("Hidden size:", model.config.n_embd)
print("Number of Transformer layers:", model.config.n_layer)
print("Number of attention heads:", model.config.n_head)

text = "Artificial Intelligence is transforming the world."

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("\nTOKENIZATION")
print("-" * 70)
print("Original text:", text)
print("\nTokens:", tokens)
print("\nToken IDs:", token_ids)

decoded_text = tokenizer.decode(token_ids)

print("\nDecoded text:")
print(decoded_text)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

prompt = "Artificial Intelligence is"

print("\n" + "=" * 70)
print("SINGLE PROMPT TEXT GENERATION")
print("=" * 70)

print("\nInput Prompt:")
print(prompt)

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    pad_token_id=tokenizer.eos_token_id
)

print("\nGenerated Text:")
print(result[0]["generated_text"])

prompts = [
    "Artificial Intelligence is",
    "The future of technology is",
    "Machine Learning allows computers to",
    "Generative AI can be used to",
    "The Internet has changed the way"
]

print("\n" + "=" * 70)
print("MULTIPLE PROMPT TEXT GENERATION")
print("=" * 70)

for prompt in prompts:
    result = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.7,
        top_k=50,
        pad_token_id=tokenizer.eos_token_id
    )

    print("\nPrompt:")
    print(prompt)

    print("Generated Text:")
    print(result[0]["generated_text"])

    print("-" * 70)

print("\n" + "=" * 70)
print("TEMPERATURE COMPARISON")
print("=" * 70)

temperature_prompt = "The future of Artificial Intelligence is"

temperatures = [0.3, 0.7, 1.0]

for temp in temperatures:
    result = generator(
        temperature_prompt,
        max_new_tokens=40,
        num_return_sequences=1,
        do_sample=True,
        temperature=temp,
        top_k=50,
        pad_token_id=tokenizer.eos_token_id
    )

    print("\nTemperature:", temp)
    print(result[0]["generated_text"])

print("\n" + "=" * 70)
print("TRANSFORMER ARCHITECTURE")
print("=" * 70)

print("""
GPT-2 is based on the Transformer architecture.

Tokens are converted into numerical representations.
Self-attention allows the model to understand relationships
between tokens.
Transformer layers process the token representations.
GPT-2 predicts the next token based on previous tokens.
The predicted tokens are repeatedly generated to produce text.
""")

print("\n" + "=" * 70)
print("OBSERVATION")
print("=" * 70)

print("""
The pretrained GPT-2 Transformer model successfully generated
text from different input prompts.

The experiment demonstrated tokenization, pretrained language
models, Transformer-based text generation, prompt-based
generation, and temperature-based generation.
""")

print("\n" + "=" * 70)
print("CONCLUSION")
print("=" * 70)

print("""
A pretrained GPT-2 Transformer language model was successfully
used for text generation. The experiment demonstrated how input
text is tokenized and processed using Transformer architecture
and self-attention to predict subsequent tokens.

GPT-2 can generate text based on a given prompt. Generation
parameters such as temperature can control the randomness and
diversity of the generated text.
""")

print("=" * 70)
print("PRACTICAL 3 COMPLETED")
print("=" * 70)
```
