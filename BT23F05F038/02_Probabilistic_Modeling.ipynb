# Practical 2 — Probabilistic Modeling and Synthetic Data

**Aim:** To implement probabilistic modeling techniques and generate synthetic data samples.

### Concepts
- Probability distributions
- Gaussian distribution
- Probability Density Function (PDF)
- Sampling
- Synthetic data generation
- Data visualization


```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import norm

np.random.seed(42)
```

## 1. Gaussian Distribution

The Gaussian (normal) probability density function is:

\[
f(x) = \frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}
\]

where:
- \(\mu\) = mean
- \(\sigma\) = standard deviation


```python
# Define Gaussian distribution parameters
mean = 50
std_dev = 10

x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 500)
pdf = norm.pdf(x, loc=mean, scale=std_dev)

plt.figure(figsize=(8, 5))
plt.plot(x, pdf)
plt.title("Gaussian Probability Density Function")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

## 2. Generate Synthetic Samples

Sampling means drawing observations from a probability distribution.

Here, we generate synthetic observations from a Gaussian distribution with:

- Mean = 50
- Standard deviation = 10


```python
# Generate synthetic data
n_samples = 1000

synthetic_data = np.random.normal(
    loc=mean,
    scale=std_dev,
    size=n_samples
)

print("Number of samples:", len(synthetic_data))
print("Sample mean:", round(np.mean(synthetic_data), 2))
print("Sample standard deviation:", round(np.std(synthetic_data), 2))
```

```python
# Visualize the generated synthetic data
plt.figure(figsize=(8, 5))
plt.hist(synthetic_data, bins=30, density=True, alpha=0.7, label="Synthetic samples")
plt.plot(x, pdf, linewidth=2, label="Theoretical Gaussian PDF")

plt.title("Synthetic Data from Gaussian Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

## 3. Generate a Two-Dimensional Synthetic Dataset

We can also model multiple features using a multivariate Gaussian distribution.


```python
# Generate 2D correlated synthetic data
mean_vector = [50, 70]

covariance_matrix = [
    [100, 60],
    [60, 144]
]

data_2d = np.random.multivariate_normal(
    mean_vector,
    covariance_matrix,
    size=1000
)

print("Shape:", data_2d.shape)
print("First five samples:")
print(data_2d[:5])
```

```python
# Visualize the 2D synthetic dataset
plt.figure(figsize=(8, 6))
plt.scatter(
    data_2d[:, 0],
    data_2d[:, 1],
    alpha=0.5,
    s=15
)

plt.title("2D Synthetic Data from a Multivariate Gaussian")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

## Conclusion

Probability distributions provide a mathematical way to model data. Once a distribution is specified, random samples can be generated from it to create synthetic datasets.

This is a basic example of probabilistic generative modeling.
