````python
# Import libraries
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
````

## Transformer Overview

A simplified transformer language-model pipeline is:

```text
Text
 ↓
Tokenization
 ↓
Token IDs
 ↓
Embeddings
 ↓
Self-Attention + Feed-Forward Layers
 ↓
Next-token probabilities
 ↓
Generated Text
```

The model predicts the next token based on the previous context.


````python
# Load a small pretrained causal language model
model_name = "distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# GPT-2 does not define a padding token by default.
tokenizer.pad_token = tokenizer.eos_token

print("Model loaded:", model_name)
````

## 1. Tokenization

Tokenization converts text into tokens that the model can process.


````python
# Example tokenization
text = "Generative artificial intelligence can"

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Original text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
````

## 2. Generate Text

The model generates one token at a time. `max_new_tokens` controls how many new tokens are generated.


````python
# Generate text
prompt = "Generative artificial intelligence can"

inputs = tokenizer(prompt, return_tensors="pt")

with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=60,
        do_sample=True,
        temperature=0.8,
        top_k=50,
        top_p=0.95,
        pad_token_id=tokenizer.eos_token_id
    )

generated_text = tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

print("Prompt:")
print(prompt)

print("\nGenerated text:")
print(generated_text)
````

## 3. Experiment with Different Prompts

Try changing the prompts below and observe how the generated output changes.


````python
prompts = [
    "Artificial intelligence is",
    "The future of machine learning is",
    "A transformer model works by",
    "Generative AI can be used for"
]

for prompt in prompts:
    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=40,
            do_sample=True,
            temperature=0.8,
            top_k=50,
            top_p=0.95,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    print("=" * 70)
    print("Prompt:", prompt)
    print("Output:", result)
````

## 4. Experiment with Temperature

Temperature affects the randomness of token selection.

- Lower temperature → more predictable output.
- Higher temperature → more diverse/random output.

Try values such as `0.3`, `0.7`, `1.0`, and `1.3`.


````python
prompt = "Machine learning is"

temperatures = [0.3, 0.7, 1.0, 1.3]

for temperature in temperatures:
    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=35,
            do_sample=True,
            temperature=temperature,
            top_k=50,
            pad_token_id=tokenizer.eos_token_id
        )

    result = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    print("=" * 70)
    print(f"Temperature: {temperature}")
    print(result)
````

## Conclusion

Transformers use self-attention to process relationships between tokens in a sequence. Pretrained transformer language models can generate text by repeatedly predicting the next token from the existing context.

The practical also demonstrates how tokenization, pretrained models, sampling, and generation parameters affect text generation.
