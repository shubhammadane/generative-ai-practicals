```python
# Practical 3: Transformer-based Language Model and Text Generation

# Install required libraries if needed:
# pip install transformers torch

from transformers import pipeline

print("Libraries imported successfully.")

# Load a pretrained GPT-2 text generation model
generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Pretrained GPT-2 model loaded successfully.")

# Input prompt
prompt = "Artificial Intelligence is changing the world because"

# Generate text
result = generator(
    prompt,
    max_length=80,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.8
)

# Display generated text
print("\nPrompt:")
print(prompt)

print("\nGenerated Text:")
print(result[0]["generated_text"])
```
