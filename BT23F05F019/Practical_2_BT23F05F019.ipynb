# 02. Probabilistic Modeling and Synthetic Data

## Aim
To implement probabilistic modeling and generate synthetic samples from a Gaussian distribution.

## Theory
A probability distribution describes uncertainty in data. The Gaussian distribution is defined by its mean and standard deviation. Sampling from the fitted distribution produces synthetic observations with similar statistical behavior.

## Expected Learning
Understand sampling, descriptive statistics, and the probability density function.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

np.random.seed(42)
real_data = np.random.normal(loc=50, scale=10, size=1000)
estimated_mean = real_data.mean()
estimated_std = real_data.std(ddof=1)
synthetic_data = np.random.normal(estimated_mean, estimated_std, size=1000)
x = np.linspace(10, 90, 500)
pdf = norm.pdf(x, estimated_mean, estimated_std)

print('Estimated mean:', round(estimated_mean, 2))
print('Estimated standard deviation:', round(estimated_std, 2))
print('Synthetic samples:', len(synthetic_data))

plt.figure(figsize=(9, 5))
plt.hist(real_data, bins=30, density=True, alpha=0.45, label='Observed data')
plt.hist(synthetic_data, bins=30, density=True, alpha=0.45, label='Synthetic data')
plt.plot(x, pdf, color='red', linewidth=2, label='Gaussian PDF')
plt.xlabel('Value')
plt.ylabel('Probability density')
plt.title('Gaussian Modeling and Synthetic Data')
plt.legend()
plt.tight_layout()
plt.show()
```

## Observation
The synthetic histogram is centered near the estimated mean and has a similar spread to the observed data. Small differences are expected because each sample is random.

## Conclusion
Thus, a Gaussian probabilistic model was fitted and used to generate synthetic data samples.
