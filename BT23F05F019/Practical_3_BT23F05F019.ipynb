# 03. Transformer-Based Language Models and Text Generation

## Aim
To study self-attention and generate text using a small transformer-style language model demonstration.

## Theory
Transformers represent tokens as vectors and use self-attention to measure how strongly each token should use information from other tokens. A language model predicts the next token from previous context. This notebook uses a tiny local corpus so it runs without downloading a pretrained model.

## Expected Learning
Understand tokens, embeddings, attention scores, and next-token generation.

```python
import numpy as np

np.random.seed(42)
corpus = 'generative ai creates new content from learned patterns. transformer models use attention to understand context. language models predict the next token.'
tokens = corpus.lower().replace('.', ' .').split()
vocabulary = sorted(set(tokens))
token_to_id = {token: i for i, token in enumerate(vocabulary)}
id_to_token = {i: token for token, i in token_to_id.items()}
embedding_size = 8
embeddings = np.random.normal(0, 1, (len(vocabulary), embedding_size))

def softmax(values):
    shifted = values - np.max(values)
    probabilities = np.exp(shifted)
    return probabilities / probabilities.sum()

def self_attention(token_ids):
    query = embeddings[token_ids[-1]]
    keys = embeddings[token_ids]
    values = embeddings[token_ids]
    scores = keys @ query / np.sqrt(embedding_size)
    weights = softmax(scores)
    context = weights @ values
    return context, weights

context_ids = [token_to_id[token] for token in ['transformer', 'models', 'use', 'attention']]
context_vector, attention_weights = self_attention(context_ids)
print('Vocabulary size:', len(vocabulary))
print('Context tokens:', [id_to_token[i] for i in context_ids])
print('Attention weights:', np.round(attention_weights, 3))
print('Context vector:', np.round(context_vector, 3))

# A transparent bigram generator demonstrates next-token generation locally.
next_tokens = {}
for current, following in zip(tokens, tokens[1:]):
    next_tokens.setdefault(current, []).append(following)
generated = ['generative']
for _ in range(12):
    choices = next_tokens.get(generated[-1], tokens)
    generated.append(choices[np.random.randint(len(choices))])
print('Generated text:', ' '.join(generated).replace(' .', '.'))
```

## Observation
Self-attention assigns different weights to the tokens in the context. The local generator produces a sequence by repeatedly selecting a likely next token from the corpus.

## Conclusion
Thus, the basic ideas behind transformer attention and autoregressive text generation were demonstrated without requiring a remote model download.
