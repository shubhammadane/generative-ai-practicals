```python
# ============================================================
# GEN AI PRACTICAL 4
# Objective: To design effective prompts and implement
# Retrieval Augmented Generation (RAG) to reduce hallucination
# in Large Language Models.
# ============================================================

# Install required libraries
!pip install -q sentence-transformers faiss-cpu

# Import libraries
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# ------------------------------------------------------------
# 1. KNOWLEDGE BASE
# ------------------------------------------------------------

documents = [
    "Python is a high-level programming language used for web development, data science and artificial intelligence.",
    "Artificial Intelligence is a field of computer science that enables machines to perform tasks that normally require human intelligence.",
    "Machine Learning is a subset of Artificial Intelligence where machines learn patterns from data.",
    "Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers.",
    "Retrieval Augmented Generation combines information retrieval with language generation to provide more accurate answers.",
    "RAG retrieves relevant information from a knowledge base and provides it as context to a language model before generating an answer."
]

print("Knowledge Base Created Successfully!")

# ------------------------------------------------------------
# 2. CREATE EMBEDDINGS
# ------------------------------------------------------------

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(documents)

print("Embeddings Created Successfully!")

# ------------------------------------------------------------
# 3. CREATE FAISS VECTOR DATABASE
# ------------------------------------------------------------

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(np.array(embeddings).astype("float32"))

print("FAISS Vector Database Created Successfully!")

# ------------------------------------------------------------
# 4. RETRIEVAL FUNCTION
# ------------------------------------------------------------

def retrieve_documents(query, k=2):

    query_embedding = model.encode([query])

    distances, indices = index.search(
        np.array(query_embedding).astype("float32"),
        k
    )

    results = []

    for i in indices[0]:
        results.append(documents[i])

    return results

# ------------------------------------------------------------
# 5. PROMPT ENGINEERING + RAG
# ------------------------------------------------------------

def rag(query):

    # Retrieve relevant documents
    retrieved_docs = retrieve_documents(query)

    # Combine retrieved documents
    context = "\n".join(retrieved_docs)

    # Create effective prompt
    prompt = f"""
You are a helpful AI assistant.

Answer the question ONLY using the information given
in the context below.

If the answer is not available in the context,
say: "I don't know based on the provided information."

Context:
{context}

Question:
{query}

Answer:
"""

    return prompt, retrieved_docs

# ------------------------------------------------------------
# 6. USER QUERY
# ------------------------------------------------------------

query = input("Enter your question: ")

prompt, retrieved_docs = rag(query)

# ------------------------------------------------------------
# 7. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n================ RAG RESULT ================\n")

print("User Question:")
print(query)

print("\nRetrieved Information:")
for doc in retrieved_docs:
    print("-", doc)

print("\nGenerated Prompt:")
print(prompt)

print("\n=============================================")
print("RAG successfully retrieved relevant information")
print("and created a context-based prompt.")
```
