# Practical 6: Image Generation using Diffusion Models (Stable Diffusion)

## Aim
To generate images using diffusion-based generative models such as Stable Diffusion.

```python
import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
import os

device = 'cuda' if torch.cuda.is_available() else 'cpu'
dtype = torch.float16 if device == 'cuda' else torch.float32

print('Using device:', device)
```

## Load Stable Diffusion Pipeline

```python
model_id = 'runwayml/stable-diffusion-v1-5'
pipe = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=dtype)
pipe = pipe.to(device)

if device == 'cuda':
    pipe.enable_attention_slicing()

print('Stable Diffusion model loaded successfully.')
```

## Generate Images from Prompts

```python
prompts = [
    'A futuristic city skyline at sunrise, ultra detailed, cinematic lighting',
    'A watercolor painting of mountains and river in monsoon season',
    'A robot reading a book in a vintage library, photorealistic'
]

negative_prompt = 'blurry, low quality, distorted, watermark, text'
images = []

for p in prompts:
    out = pipe(
        prompt=p,
        negative_prompt=negative_prompt,
        num_inference_steps=30,
        guidance_scale=7.5
    )
    images.append(out.images[0])

print('Image generation completed.')
```

```python
fig, axes = plt.subplots(1, len(images), figsize=(18, 6))
for i, img in enumerate(images):
    axes[i].imshow(img)
    axes[i].set_title(f'Prompt {i+1}')
    axes[i].axis('off')
plt.tight_layout()
plt.show()
```

```python
os.makedirs('outputs_practical6', exist_ok=True)
for i, img in enumerate(images, start=1):
    path = f'outputs_practical6/generated_{i}.png'
    img.save(path)
    print('Saved:', path)
```

## Result
Diffusion-based image generation was implemented successfully using Stable Diffusion and prompt-based synthesis.
